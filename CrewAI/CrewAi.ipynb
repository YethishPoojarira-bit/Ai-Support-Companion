````python
# 1. Configure the LLM (Azure OpenAI via LiteLLM)
# Use azure/ prefix with is_litellm=True to use LiteLLM for Azure OpenAI
llm = LLM(
    model="azure/gpt-4o-mini",  # azure/<deployment-name>
    temperature=0.2,
    api_key=os.getenv("AZURE_OPENAI_API_KEY"),
    base_url=os.getenv("AZURE_OPENAI_ENDPOINT"),
    api_version="2024-12-01-preview",
    is_litellm=True  # Force using LiteLLM
)

print("LLM configured successfully with Azure OpenAI via LiteLLM.")
print(llm.call("hello, how are you?"))
````

# **2. Building Email Assistant Agent**

````python
# 1. Build the Email Agent
Email_Agent = Agent(
    name = "Email Agent",
    role = "An AI agent that manages and organizes emails efficiently.",
    goal = "improve email make it more professional, clear and concise",
    backstory = "You are an expert in email communication skilled in professional email writing.",
    verbose = True,
    llm = llm
)

# 2. Create Tasks for the Email Agent
Mail_Task = Task(
    name = "Draft Professional Email",
    description = "Compose a professional email to a client regarding project updates.",
    parameters = {
        "recipient": "",
        "subject": "",
        "body": ""
    },
    expected_output="A professional email draft",
    agent = Email_Agent
)

# 3. Initialize the Crew with the Email Agent and Tasks
email_crew = Crew(
    agents = [Email_Agent],
    tasks = [Mail_Task],
    llm = llm,
    verbose = False
)

email_crew.kickoff(inputs={
    "recipient": "projectwork@gmail.com",
    "subject": "Project Update",
    "body": "I wanted you to provide an update on the project status."
})
````

### Explanation of the Code

#### **1. What is an `Agent`?**
An `Agent` in CrewAI is an AI entity designed to perform specific roles or tasks. It encapsulates the logic, behavior, and personality of the AI for a particular purpose.

- **Why does it have a `name`?**
  The `name` is a human-readable identifier for the agent. It describes the agent's role or purpose (e.g., "Email Agent"). This is different from the variable name (`Email_Agent`), which is just a Python variable used to reference the object in your code.

  - **Example:**
    - `name="Email Agent"`: Describes the agent's purpose.
    - `Email_Agent`: The Python variable used to reference the agent object.

- **Other attributes:**
  - `role`: Describes what the agent does (e.g., "manages and organizes emails efficiently").
  - `goal`: The agent's objective (e.g., "improve email, make it more professional, clear, and concise").
  - `backstory`: Adds context to the agent's expertise or personality.
  - `llm`: The language model (e.g., OpenAI or Gemini) that powers the agent's intelligence.

---

#### **2. What is a `Task`?**
A `Task` represents a specific job or action that the agent is expected to perform. It defines the scope of work, the input parameters, and the expected output.

- **Why does it have a `name`?**
  The `name` is a human-readable identifier for the task (e.g., "Draft Professional Email"). It describes what the task is about. This is different from the variable name (`Mail_Task`), which is just a Python variable used to reference the task object.

  - **Example:**
    - `name="Draft Professional Email"`: Describes the task's purpose.
    - `Mail_Task`: The Python variable used to reference the task object.

- **Why do we need to specify the `agent` in the task?**
  The `agent` field explicitly links the task to the agent responsible for performing it. This is necessary because a `Crew` can have multiple agents, and each task must be assigned to a specific agent.

  - **Example:**
    If you have multiple agents (e.g., "Email Agent" and "Calendar Agent"), you need to specify which agent is responsible for each task.

---

#### **3. What is `verbose`?**
The `verbose` parameter controls the level of detail in the output logs. When `verbose=True`, the system provides detailed logs about the agent's actions, decisions, and task execution. This is useful for debugging or understanding how the agent is working.

- **When to use it:**
  - `verbose=True`: Use when you want detailed logs for debugging or monitoring.
  - `verbose=False`: Use when you want minimal output (e.g., in production).

---

#### **4. Difference Between `kickoff` and `start`**
Both `kickoff` and `start` are methods used to initiate the execution of tasks, but they differ in their behavior:

- **`kickoff`:**
  - Used to start the execution of tasks with **initial inputs** provided by the user.
  - You pass a dictionary of `inputs` to provide the necessary parameters for the task.
  - Example:
    ```python
    email_crew.kickoff(inputs={
        "recipient": "projectwork@gmail.com",
        "subject": "Project Update",
        "body": "I wanted you to provide an update on the project status."
    })
    ```

- **`start`:**
  - Used to start the execution of tasks **without requiring initial inputs**.
  - Typically used when the task does not require user-provided inputs or when inputs are already pre-configured.

- **Key Difference:**
  - `kickoff` is for tasks that require user-provided inputs at runtime.
  - `start` is for tasks that can run without additional inputs.

---

### Summary
- **Agent**: Represents the AI entity with a specific role and purpose. The `name` describes its purpose, while the variable name is just a Python reference.
- **Task**: Represents a specific job for the agent. The `name` describes the task, and the `agent` links the task to the responsible agent.
- **Verbose**: Controls the level of logging detail.
- **Kickoff vs Start**: `kickoff` is for tasks requiring user inputs, while `start` is for tasks that don't.

# **3. Using Tools**

````python
# Creating a custom tool using the @tool decorator from crewai.tools
from crewai.tools import tool

@tool
def word_correction_tool(text: str) -> str:
    """Corrects common text shortcuts and informal words to their full forms. 
    Use this tool to expand shortcuts like 'u', 'r', 'btw', 'idk', 'omg', 'ppl' into proper words."""
    corrections = {
        "u": "you",
        "r": "are",
        "btw": "by the way",
        "idk": "I don't know",
        "omg": "oh my god",
        "ppl": "people"
    }
    
    import re
    result = text
    for shortcut, full_form in corrections.items():
        # Use word boundaries to avoid replacing parts of words
        result = re.sub(r'\b' + shortcut + r'\b', full_form, result, flags=re.IGNORECASE)
    
    return result

# Test the tool
print(word_correction_tool.run("omg u r amazing btw idk what to say"))
````

````python
# 1. Build the Email Agent
Email_Rewrite_Agent = Agent(
    name = "Email Rewrite Agent",
    role = "An AI agent that rewrites emails to be more professional and friendly",
    goal = "Rewrite the given email subject and body to make them professional, friendly, clear and concise",
    backstory = "You are an expert in email communication skilled in professional email writing. You excel at transforming casual or informal messages into polished, professional communications.",
    verbose = False,
    llm = llm,
    tools = [word_correction_tool]  # Add tools to the agent
)

# 2. Create Tasks for the Email Agent
Rewrite_Mail_Task = Task(
    name = "Rewrite Email",
    description = """
    First, use the Word Correction Tool to fix any shortcuts or informal language in the subject and body.
    Then, rewrite the email to make it professional and friendly.
    
    Given inputs:
    - Recipient: {recipient}
    - Subject: {subject}
    - Body: {body}
    
    Your task:
    1. Apply word corrections to fix shortcuts (u -> you, r -> are, etc.)
    2. Rewrite the subject to be professional and clear
    3. Rewrite the body to be professional, friendly, and well-structured
    4. Format as a complete email with proper greeting and closing
    """,
    expected_output="A complete professional email draft",
    agent = Email_Rewrite_Agent
)

# Create a Crew
email_crew = Crew(
    agents = [Email_Rewrite_Agent],
    tasks = [Rewrite_Mail_Task],
    llm = llm,
    verbose=False
)

# Execute the crew
result = email_crew.kickoff(inputs={
   "recipient" : "team@gmail.com",
   "subject" : "just to know how are u ppl",
   "body" : "hello, guys how u all doing hope u r all fine, got a surprise for u idk why but u all got a holiday tomorrow, omg have fun" 
})

print("\n" + "="*80)
print("FINAL RESULT:")
print("="*80)
print(result)
````

# **4. Dependent Agent.**

````python
#creat a agent to perform research on a topic and summarize findings
research_agent = Agent(
    name = "Research Agent",
    role = "Research and gather information on specified topics to support decision-making processes.",
    goal = "Conduct thorough research and provide accurate, concise summaries of findings.",
    backstory = "You are a skilled researcher with expertise in data collection, analysis, and synthesis. You excel at finding reliable sources and distilling complex information into clear, actionable insights.",
    llm = llm
)

#creat a agent to write blog posts on various topics
blog_writer_agent = Agent(
    name = "Blog Writer Agent",
    role = "Create engaging and informative blog posts on a variety of topics.",
    goal = "Write well-structured blog posts of 1000 words that captivate readers and provide valuable information.",
    backstory = "You are a talented writer with a knack for crafting compelling blog content. You understand how to engage readers and communicate ideas effectively.",
    llm = llm
)
````

````python
research_task = Task(
    name = "Research Topic",
    description = '''
    - Conduct research on the given topic: {topic} and provide a summary of findings.
    - fetch around 10 points on {topic}.
    - Ensure the information is accurate, relevant, and well-organized.
    ''',
    expected_output="A concise research pointwise summary with key facts and insights",
    agent = research_agent,
    parameters = {
        "topic": ""
    }
)

summarize_data = Task(
    name = "Summarize Data",
    description = "Summarize the research data collected on the topic: {topic}.",
    expected_output="A clear and concise summary of the research data within 100 words",
    agent = research_agent,
    context=[research_task]  # This automatically provides research_task output to this task
)

blog_writer_task = Task(
    name = "Write Blog Post",
    description = """Based on the research findings, write a comprehensive 100-word blog post on {topic}.
    
    Use the research summary from the previous task to ensure accuracy and incorporate key insights.
    Structure the blog post with:
    - An engaging introduction
    - Well-organized main sections with subheadings
    - A compelling conclusion
    """,        
    expected_output="A well-structured 100-word blog post based on the research",
    agent = blog_writer_agent,
    context=[research_task]  # This automatically provides research_task output to this task
)
````

````python
blog_crew = Crew(
    agents = [research_agent, blog_writer_agent],
    tasks = [research_task, summarize_data, blog_writer_task],
    llm = llm,
    verbose=True
)

blog_crew.kickoff(inputs={
    "topic": "The Impact of Artificial Intelligence on Modern Workplaces"
})
````

````python
from crewai import Agent, Crew, Task
from crewai.utilities.prompts import Prompts

# Create your agent
agent = Agent(
    role="Data Analyst",
    goal="Analyze data and provide insights",
    backstory="You are an expert data analyst with 10 years of experience.",
    verbose=True,
    llm = llm
)

# Create a sample task
task = Task(
    description="Analyze the sales data and identify trends",
    expected_output="A detailed analysis with key insights and trends",
    agent=agent
)

# Create the prompt generator
prompt_generator = Prompts(
    agent=agent,
    has_tools=len(agent.tools) > 0,
    use_system_prompt=agent.use_system_prompt
)

# Generate and inspect the actual prompt
generated_prompt = prompt_generator.task_execution()

# Print the complete system prompt that will be sent to the LLM
if "system" in generated_prompt:
    print("=== SYSTEM PROMPT ===")
    print(generated_prompt["system"])
    print("\n=== USER PROMPT ===")
    print(generated_prompt["user"])
else:
    print("=== COMPLETE PROMPT ===")
    print(generated_prompt["prompt"])

# You can also see how the task description gets formatted
print("\n=== TASK CONTEXT ===")
print(f"Task Description: {task.description}")
print(f"Expected Output: {task.expected_output}")
````
